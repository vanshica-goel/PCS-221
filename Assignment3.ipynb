```python
#import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
## Step 1: Initialize Random Seed
ROLL_NUMBER = 108
np.random.seed(ROLL_NUMBER)
```

```python
## Step 2: Generate Sales Data
sales_data = np.random.randint(1000, 5001, size=(12, 4))
sales_data
```

```python
## Step 3: Convert NumPy Array to Pandas DataFrame
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

columns = ["Electronics", "Clothing", "Home & Kitchen", "Sports"]

df = pd.DataFrame(sales_data, index=months, columns=columns)
df
```

```python
## Part2
## Display First 5 Rows and Summary Statistics
df.head()
df.describe()
```

```python

# Total sales per category
total_category = df.sum()
total_category
```

```python
# Total sales per month
total_month = df.sum(axis=1)
total_month
```

```python
## Calculate Average Sales Growth Between Consecutive Months
growth = df.pct_change()
average_growth = growth.mean()
average_growth
```

```python
## Add Total Sales and Growth Rate Columns
df["Total Sales"] = df.sum(axis=1)
df["Growth Rate"] = df["Total Sales"].pct_change() * 100

df
```

```python
## Apply Randomized Discount
if ROLL_NUMBER % 2 == 0:
    df["Electronics"] = df["Electronics"] * 0.90
    print("10% discount applied to Electronics")
else:
    df["Clothing"] = df["Clothing"] * 0.85
    print("15% discount applied to Clothing")

df
```

```python
## part3 visualization
## Monthly Sales Trends (Line Plot)
plt.figure(figsize=(10,6))

plt.plot(df.index, df["Electronics"], marker='o')
plt.plot(df.index, df["Clothing"], marker='o')
plt.plot(df.index, df["Home & Kitchen"], marker='o')
plt.plot(df.index, df["Sports"], marker='o')

plt.title("Monthly Sales Trends by Category")
plt.xlabel("Month")
plt.ylabel("Sales Units")
plt.legend(["Electronics", "Clothing", "Home & Kitchen", "Sports"])
plt.xticks(rotation=45)

plt.show()
```

```python
## Box Plot for Sales Distribution
plt.figure(figsize=(8,5))

sns.boxplot(data=df[["Electronics", "Clothing", "Home & Kitchen", "Sports"]])

plt.title("Sales Distribution by Category")
plt.ylabel("Sales Units")

plt.show()
```

```python
## 2
import numpy as np

array = np.array([[1, -2, 3],
                  [-4, 5, -6]])
```

```python
absolute_array = np.abs(array)
absolute_array
```

```python
## Percentile of Flattened Array
flattened = array.flatten()

p25_flat = np.percentile(flattened, 25)
p50_flat = np.percentile(flattened, 50)
p75_flat = np.percentile(flattened, 75)

p25_flat, p50_flat, p75_flat
```

```python
##Percentile for Each Column
p25_col = np.percentile(array, 25, axis=0)
p50_col = np.percentile(array, 50, axis=0)
p75_col = np.percentile(array, 75, axis=0)

p25_col, p50_col, p75_col
```

```python
##Percentile for Each row
p25_row = np.percentile(array, 25, axis=1)
p50_row = np.percentile(array, 50, axis=1)
p75_row = np.percentile(array, 75, axis=1)

p25_row, p50_row, p75_row
```

```python
## Mean, Median and Standard Deviation of flattened array
mean_flat = np.mean(flattened)
median_flat = np.median(flattened)
std_flat = np.std(flattened)

mean_flat, median_flat, std_flat
```

```python
##for each column
mean_col = np.mean(array, axis=0)
median_col = np.median(array, axis=0)
std_col = np.std(array, axis=0)

mean_col, median_col, std_col
```

```python
##for each row
mean_row = np.mean(array, axis=1)
median_row = np.median(array, axis=1)
std_row = np.std(array, axis=1)

mean_row, median_row, std_row
```

```python
##4
# Create a list
my_list = [10, 20, 30, 40, 50]

# Positions to swap
pos1 = 1   # index of 20
pos2 = 3   # index of 40

print("Original List:", my_list)

# Swapping using temporary variable
temp = my_list[pos1]
my_list[pos1] = my_list[pos2]
my_list[pos2] = temp

print("List after swapping:", my_list)
```

```python
#5 Original set
my_set = {10, 20, 30, 40, 50}

print("Original Set:", my_set)

# Convert set to list
my_list = list(my_set)

# Positions to swap
pos1 = 1
pos2 = 3

# Swapping using temporary variable
temp = my_list[pos1]
my_list[pos1] = my_list[pos2]
my_list[pos2] = temp

# Convert back to set
my_set = set(my_list)

print("Set after swapping:", my_set)
```
